```python
# Examples --> Unique pairs
for i in range(1,3):
    for j in range(1,3):
        print(i,j)
```

```python
row = int(input("Enter the number of rows: "))
for i in range(1,row+1):
    for j in range(1,i+1):
        print('*', end =' ')
    print()
```

```python
num = int(input('Enter you number:'))
for i in range(1,num+1):
    for j in range(1, i+1):
        print(j, end =' ')
    for k in range(i-1, 0, -1):
        print(k, end=' ')
    print()
```
